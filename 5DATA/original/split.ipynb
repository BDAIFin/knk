```python
import pandas as pd
trans = pd.read_csv("transactions_data.csv")
```

```python
train = trans[trans["date"] < "2016-01-01"].copy()
test  = trans[(trans["date"] >= "2016-01-01") & (trans["date"] < "2017-01-01")].copy()
check = trans[trans["date"] >= "2017-01-01"].copy()
```

```python
check.to_parquet("check")
```

```python
train.to_parquet("train")
test.to_parquet("test")
check.to_parquet("check")
```
